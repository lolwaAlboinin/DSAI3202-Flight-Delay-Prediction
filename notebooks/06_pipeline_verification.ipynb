# Pipeline Verification — v1 and v2

Cell 1 checks registered v1 tables; Cell 2 checks registered production v2 tables.

```python

from pyspark.sql import functions as F

schema = "dsai3202_databricks.proj_flight60106354"

clean = spark.table(f"{schema}.flights_clean_2025")
features = spark.table(f"{schema}.flights_features_2025")

clean_count = clean.count()
feature_count = features.count()

assert clean_count == 6879484
assert feature_count == clean_count

assert "ArrDelay" not in features.columns

invalid_targets = features.filter(
    F.col("ArrDel15").isNull() |
    (~F.col("ArrDel15").isin(0, 1))
).count()

assert invalid_targets == 0

delay_stats = features.agg(
    F.sum("ArrDel15").alias("Delayed"),
    F.count("*").alias("Total")
).first()

print("PIPELINE VERIFICATION RESULTS")
print("Clean records:", clean_count)
print("Feature records:", feature_count)
print("Delayed records:", delay_stats["Delayed"])
print("Delay rate:",
      round(100 * delay_stats["Delayed"] / delay_stats["Total"], 2), "%")
print("PASS: All publication checks completed")
```
